## Section 1: My Lane as an ML Task

### Chosen Lane
**Search Intent & High-Intent Lead Conversion Scoring**

### ML Task Type
**Binary Classification / Intent Scoring**

### Description
This task classifies incoming user sessions and content interactions into **High-Intent Lead (1)** vs. **Low-Intent / General Visitor (0)** based on interaction patterns, session duration, depth of page visits, and keyword query attributes.

## Section 2: Target or Proxy Variable

### Target Variable (`is_high_intent`)
* **Type:** Binary Target (0 or 1)
* **Definition:**  
  * `1` = Visitor triggers a primary conversion action (e.g., clicks 'Book Discovery Call', submits inquiry, or views >3 technical case studies with dwell time > 120s).
  * `0` = Visitor bounces or browses passively without engaging with core conversion elements.

## Section 3: Success Metric

### Selected Metrics
1. **F1-Score (Primary):** Balances Precision and Recall, ensuring we don't spam low-intent leads while minimizing missed high-value opportunities.
2. **Precision @ K:** Ensures that the top 10% scored leads presented to sales/automation workflows are truly high intent.

## Section 4: Unit of Analysis (Real Dataframe)

**Definition:** **One row = One unique visitor session / interaction payload.**

## Section 5: Why ML Beats a Fixed Rule Here

1. **Non-Linear Interactions:** Fixed heuristic rules (e.g., `dwell_time > 100s AND pages > 3`) fail to capture nuanced behavior (e.g., a user coming from LinkedIn who reads 1 deep case study for 90s is higher intent than a user who clicks around 5 pages passively).
2. **Continuous Adaptability:** Fixed rules decay over time as traffic sources change. ML models dynamically adapt weights based on updated conversion feedback.
3. **Probabilistic Scoring:** ML provides a calibrated probability score (0.00 to 1.00) rather than a rigid pass/fail toggle, allowing automated systems to prioritize lead routing dynamically.
